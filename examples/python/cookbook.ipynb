````python
import inference_rs as ir
from inference_rs import types as t

engine = ir.Engine(
    t.EngineSpec(
        model=t.ModelSelectedGGUF(
            tok_model_id="mistralai/Mistral-7B-Instruct-v0.1",
            quantized_model_id="TheBloke/Mistral-7B-Instruct-v0.1-GGUF",
            quantized_filename="mistral-7b-instruct-v0.1.Q4_K_M.gguf",
        )
    )
)

res = engine.chat(
    t.ChatCompletionRequest(
        model="default",
        messages=[
            t.Message(
                role="user", content="Tell me a story about the Rust type system."
            )
        ],
        max_tokens=256,
        presence_penalty=1.0,
        top_p=0.1,
        temperature=0.1,
    )
)
print(res)
````

This imports the package and its generated types. `ir.Engine` is a class which handles loading and running the model, and the `t.EngineSpec` it takes says which model to load, through one of the `t.ModelSelected*` classes.

This tells the `Engine` to actually load the model. It will use a CUDA, Metal, or CPU device depending on what `features` you set during compilation: [here](https://github.com/christopherthompson81/inference.rs?tab=readme-ov-file#supported-accelerators).

Now we actually send a request! We can specify the messages just like with an OpenAI API.

````python
import inference_rs as ir
from inference_rs import types as t

engine = ir.Engine(
    t.EngineSpec(
        model=t.ModelSelectedPlain(
            model_id="mistralai/Mistral-7B-Instruct-v0.1",
            arch=t.NormalLoaderType.MISTRAL,
        )
    )
)

res = engine.chat(
    t.ChatCompletionRequest(
        model="default",
        messages=[
            t.Message(
                role="user", content="Tell me a story about the Rust type system."
            )
        ],
        max_tokens=256,
        presence_penalty=1.0,
        top_p=0.1,
        temperature=0.1,
    )
)
print(res)
````

Lets walk through this code too and see the difference between loading a Plain model and loading a GGUF model.
```python
import inference_rs as ir
from inference_rs import types as t
```

This imports the package and its generated types. `ir.Engine` is a class which handles loading and running the model, and the `t.EngineSpec` it takes says which model to load, through one of the `t.ModelSelected*` classes. Note that we also use the `t.NormalLoaderType` enum which controls the non-GGUF model's architecture.

```python
engine = ir.Engine(

    )
)
```

This tells the `Engine` to actually load the model as a Mistral architecture. It will use a CUDA, Metal, or CPU device depending on what `features` you set during compilation: [here](https://github.com/christopherthompson81/inference.rs?tab=readme-ov-file#supported-accelerators).

## Loading a Mistral + GGUF model

````python
import inference_rs as ir
from inference_rs import types as t

engine = ir.Engine(
    t.EngineSpec(
        model=t.ModelSelectedGGUF(
            tok_model_id="mistralai/Mistral-7B-Instruct-v0.1",
            quantized_model_id="TheBloke/Mistral-7B-Instruct-v0.1-GGUF",
            quantized_filename="mistral-7b-instruct-v0.1.Q4_K_M.gguf",
        )
    )
)
````

## Loading a plain Mistral model

````python
import inference_rs as ir
from inference_rs import types as t

engine = ir.Engine(
    t.EngineSpec(
        model=t.ModelSelectedPlain(
            model_id="mistralai/Mistral-7B-Instruct-v0.1",
            arch=t.NormalLoaderType.MISTRAL,
        )
    )
)
````

## Running the Engine

````python
from inference_rs import types as t

res = engine.chat(
    t.ChatCompletionRequest(
        model="default",
        messages=[
            t.Message(
                role="user", content="Tell me a story about the Rust type system."
            )
        ],
        max_tokens=256,
        presence_penalty=1.0,
        top_p=0.1,
        temperature=0.1,
    )
)
print(res.choices[0].message.content)
print(res.usage)
````
